```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../results/exp1_dynamics.csv")
```

```python
sns.lineplot(data=df, x="epoch", y="R_S_shuffle", hue="seed", marker="o")
plt.axhline(0.5, ls="--", color="gray")
plt.title("R_S_shuffle over training (Exp 1)")
plt.ylim(0, 1.05)
plt.show()
```

```python
sns.lineplot(data=df, x="epoch", y="R_S_conflict", hue="seed", marker="o")
plt.axhline(0.5, ls="--", color="gray")
plt.title("R_S_conflict over training (Exp 1)")
plt.ylim(0, 1.05)
plt.show()
```
