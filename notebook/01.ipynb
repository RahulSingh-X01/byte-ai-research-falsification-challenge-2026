```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../results/exp0.csv")
df["config"] = df.apply(
    lambda r: f"S" if r["feature"] == "S" else f"C k={int(r['k'])}", axis=1
)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.barplot(data=df, x="config", y="test_acc", ax=axes[0])
axes[0].set_ylim(0.9, 1.01)
axes[0].set_title("Test accuracy (Exp 0)")
sns.barplot(data=df, x="config", y="epochs_to_95", ax=axes[1])
axes[1].set_title("Epochs to 95% train accuracy (Exp 0)")
plt.tight_layout()
plt.show()
```
